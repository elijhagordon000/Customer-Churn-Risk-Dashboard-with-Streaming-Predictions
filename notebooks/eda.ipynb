```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv')

# Target Colunmn:
# Churn

# Numeric Columns:
# SeniorCitizen(int64)
# tenure(int64)
# MonthlyCharges(float64)
# TotalCharges(object)

# Categorical Columns(All object types):
# customerID
# gender
# Partner
# Dependents
# PhoneService
# MultipleLines
# InternetService
# OnlineSecurity
# OnlineBackup
# DeviceProtection
# TechSupport
# StreamingTV
# StreamingMovies
# Contract
# PaperlessBilling
# PaymentMethod
# TotalCharges is currently text but should likely be converted to numeric
# Churn

# Clening Data:

# # print(df.to_string())
# print('df.columns')
# print(df.columns)
# print('Head:')
# print(df.head)
# print('df.dtypes:')
# print(df.dtypes)

# Checking for missing values:
# df.isnull().sum()
# Checking for blank strings
# (df == "").sum()
# Checking for cells that are spaces
# (df.astype(str).apply(lambda col: col.str.strip() == "")).sum()
# inspecting TotalCharges because there are 11 rows where cells are just spaces
# df[df["TotalCharges"].astype(str).str.strip() == ""]

# Replacing blank TotalCharges with 0 and converting TotalCharges to numeric type:
# Making sure that it converts all types of blank spaces to "":
df["TotalCharges"] = df["TotalCharges"].astype(str).str.strip()
df["TotalCharges"] = df["TotalCharges"] = df["TotalCharges"].replace("", "0")
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])
# df["TotalCharges"].dtype output: dtype('float64')

# Checking Churn Balance
# print(df["Churn"].value_counts())
# print(df["Churn"].value_counts(normalize=True))

# Displaying Customer Churn Count:
df["Churn"].value_counts().plot(kind="bar")
plt.title("Churn Count")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()

# Displaying Churn By Contract Type:
pd.crosstab(df["Contract"], df["Churn"]).plot(kind="bar")
plt.title("Churn by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.show()

# Displaying Churn By Internet Service:
pd.crosstab(df["InternetService"], df["Churn"]).plot(kind="bar")
plt.title("Churn by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Number of Customers")
plt.show()

# Displaying Churn by Monthly Charges distribution
df["MonthlyCharges"].plot(kind="hist", bins=30)
plt.title("Distribution of Monthly Charges")
plt.xlabel("Monthly Charges")
plt.ylabel("Frequency")
plt.show()


```
